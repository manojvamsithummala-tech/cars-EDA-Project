```python
#Outlier Treatment
sns.boxplot(df[['Model_Year','Engine_Size','Mileage','Doors','Horsepower','Price']])
plt.title("Outlier detection")
plt.show()
```

```python
#Outlier treatment
for col in ['Price']:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3-Q1
  lower = Q1-1.5*IQR
  upper = Q3+1.5*IQR
  df[col] = np.where(df[col]>upper,upper,np.where(df[col]<lower,lower,df[col]))
```

```python
#Outlier Treatment
sns.boxplot(df[['Price']])
plt.title("Outlier detection")
plt.show()
```

```python
# Univariate Analysis
sns.displot(df['Model_Year'])
plt.show()
```
